# Comparing eight bankruptcy classifiers

Comparison of LDA, logistic regression, decision trees, random forest, SVM, KNN, naive Bayes and gradient boosting.

Saved outputs are from the original coursework run, before the small fixes in this copy. The notebooks have not been rerun.


```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import GradientBoostingClassifier

# Load dataset into a Pandas DataFrame
df = pd.read_excel('data/business_data_encrypted.xlsx')

# Drop the specified columns
df = df.drop(df.columns[0:3], axis=1)
df = df.drop(df.columns[1:27], axis=1)
#print("\n\n", df.head(), "\n\n")

# Extract the target variable
y = df['label']

# Preserve the original feature selection; the first remaining column is the label.
scaler = StandardScaler()
X = df.drop(df.columns[0], axis=1)

# Splitting the dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Applying LDA
print("0: Linear Discriminant Analysis")
model = LinearDiscriminantAnalysis()
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Logistic Regression Classifier
print("1: Logistic Regresion")
model = LogisticRegression()
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Decision Trees Classifier
print("2: Decision Trees")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Random Forest Classifier
print("3: Random Forest")
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Support Vector Machines Classifier
print("4: Support Vector Machines")
model = SVC()
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
X_train_scaled = scaler.fit_transform(X_train.values)
X_test_scaled = scaler.transform(X_test.values)

# Applying K-Neighbors Classifier
print("5: K-Neighbors Classifier")
model = KNeighborsClassifier(n_neighbors=2)
model.fit(X_train_scaled, y_train)

# Predicting on the test set
y_pred = model.predict(X_test_scaled)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Naive Bayes Classifier
print("6: Naive Bayes")
model = GaussianNB()
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```

```python
# Applying Gradient Boosting Classifier
print("7: Gradient Boosting Classifier")
model = GradientBoostingClassifier(random_state=42)
model.fit(X_train, y_train)

# Predicting on the test set
y_pred = model.predict(X_test)

# Computing the confusion matrix
conf_matrix_more_complex = confusion_matrix(y_test, y_pred)

# Generating the classification report
report = classification_report(y_test, y_pred)

# Print the confusion matrix and the classification report
print("Confusion Matrix:")
print(conf_matrix_more_complex)
print("\nClassification Report:")
print(report)
```
