# Rosenbrock Hessian check

Searches a grid for a Hessian with a negative eigenvalue as a counterexample to convexity.

Saved outputs are from the original coursework run, before the small fixes in this copy. The notebooks have not been rerun.


```python
import numpy as np
import numdifftools as nd
from scipy.optimize import minimize


#def f(x):
    # Example multivariable function: f(x, y) = x^2 + y^2
 #   return x[0]**2 + x[1]**2
def f(x, a=1, b=100):
    #Rosenbrock function
    return (a - x[0])**2 + b*(x[1] - x[0]**2)**2

def is_positive_semidefinite(H):
    return np.all(np.linalg.eigvals(H) >= 0)

def objective_function(x):
    return x[0]**2 + x[1]**2


# Check convexity in a range
start, end = -10, 10
step = 1
convex = True

for x in np.arange(start, end, step):
    for y in np.arange(start, end, step):
        H = nd.Hessian(f)([x, y])
        if not is_positive_semidefinite(H):
            convex = False
            break

    if not convex:
        break

print("The function is convex:", convex)
```
